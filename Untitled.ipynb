```python
subj = f._subjects["subj_0"]
X = subj.matrix()["rel"].T
dXdt = np.diff(X, axis=0)/np.diff(subj.times).reshape((-1,1))

X = X[:-1]


for i in range(1, 10):
    subj = f._subjects[f"subj_{i}"]
    x = subj.matrix()["rel"].T
    dxdt = np.diff(x, axis=0)/np.diff(subj.times).reshape((-1,1))

    X = np.concatenate([X, x[:-1]], axis=0)
    dXdt = np.concatenate([dXdt, dxdt], axis=0)
    
```

```python
reg = lm.LinearRegression(fit_intercept=False)
```

```python
p_est = np.zeros((3, 4))

for i in range(3):
    reg.fit(np.hstack([X[:,[i]], X[:,[i]]*X]), dXdt[:,i])
    p_est[i] = reg.coef_
```

```python
def extract_interactions(_mcmc: md2.BaseMCMC):
    interactions = _mcmc.graph[STRNAMES.INTERACTIONS_OBJ].get_trace_from_disk(section='posterior')

    self_interactions = _mcmc.graph[STRNAMES.SELF_INTERACTION_VALUE].get_trace_from_disk(section='posterior')
    self_interactions = -np.absolute(self_interactions)

    for i in range(self_interactions.shape[1]):
        interactions[:, i, i] = self_interactions[:, i]
    return interactions


def extract_growths(_mcmc: md2.BaseMCMC):
    return _mcmc.graph[STRNAMES.GROWTH_VALUE].get_trace_from_disk(section='posterior')
```

```python
def extract_interactions2(tracer):
    interactions = np.array(tracer[STRNAMES.INTERACTIONS_OBJ])

    self_interactions = np.array(tracer[STRNAMES.SELF_INTERACTION_VALUE])
    self_interactions = -np.absolute(self_interactions)

    for i in range(self_interactions.shape[1]):
        interactions[:, i, i] = self_interactions[:, i]
    return interactions
```
